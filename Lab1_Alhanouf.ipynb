```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
for mark in marks:
  if mark >= 60:
    print(mark, "Pass")
  else:
    print(mark, "Fail")
```

```python
print("Number of samples:", len(marks))
average_hours = sum(study_hours) / len(study_hours)
print("Average study hours:", average_hours)
average_mark = sum(marks) / len(marks)
print("Average mark:", average_mark)
print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))
print("\nStudy Hours / Mark:")
for i in range(len(marks)):
    print(study_hours[i], "hours:", marks[i])
print("\nPass/Fail:")
for mark in marks:
    if mark >= 60:
        print(mark,":Pass")
    else:
        print(mark,":Fail")
passes = 0
for mark in marks:
    if mark >= 60:
        passes += 1
print("\nNumber of students who passed:", passes)

```
