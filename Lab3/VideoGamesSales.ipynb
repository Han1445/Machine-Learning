```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
```

```python
df = pd.read_csv("Video_Games_Sales_as_at_22_Dec_2016.csv")
df.head(-1)
```

```python
df.dtypes
```

```python
df['User_Score'] = pd.to_numeric(df['User_Score'], errors='coerce')
df.dtypes
```

```python
print(df.isna().sum())
```

```python
df.duplicated()[df.duplicated()==True]
```

```python
df.describe(include='all')
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Critic_Score'], bins=20)
plt.title("Distribution of Critic Scores")
plt.show()
```

```python
platform_sales = df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,5))
platform_sales.plot(kind='bar')
plt.title("Total Global Sales by Platform")
plt.ylabel("Global Sales")
plt.show()
platform_sales
```

```python
genre_sales = df.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,5))
genre_sales.plot(kind='bar')
plt.title("Total Global Sales by Genre")
plt.ylabel("Global Sales")
plt.xlabel("Genre")
plt.show()
genre_sales
```

```python
publisher_sales = df.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False)
publisher_sales.head(10)
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Critic_Score', y='Global_Sales', data=df)
plt.title("Critic Score vs Global Sales")
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
yearly_sales = df.groupby('Year_of_Release')['Global_Sales'].sum()
plt.figure(figsize=(10,5))
yearly_sales.plot()
plt.title("Global Sales Trend by Year")
plt.ylabel("Global Sales")
plt.xlabel("Year")
plt.show()
```
